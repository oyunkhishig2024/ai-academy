```python
def add_numbers(*numbers):
    return sum(numbers)

print(add_numbers(1,2,3,4,5))
print(add_numbers(1,2,3))
```

```python
def multiply_many(*numbers):
    total = 1
    num = list(numbers)
    for i in num:
        total *= i
    return total
print(multiply_many(1,2,3,4,5))
```
